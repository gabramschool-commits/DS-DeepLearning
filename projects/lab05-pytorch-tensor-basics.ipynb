## Task 1 — Perform Standard Imports

We import **NumPy** (used to create the initial random array) and **PyTorch** (used for all tensor operations).

```python
import numpy as np
import torch

print("NumPy version:  ", np.__version__)
print("PyTorch version:", torch.__version__)
```

---
## Task 2 — Create the `set_seed()` Function

This function sets the random seed for both NumPy and PyTorch so that any random values generated afterward are reproducible. Both libraries maintain separate random number generators, so both need to be seeded.

```python
def set_seed(seed: int) -> None:
    """Set the random seed for NumPy and PyTorch."""
    np.random.seed(seed)
    torch.manual_seed(seed)
```

The function accepts an integer seed, sets it for both libraries, and returns nothing. Calling `set_seed(42)` before generating random values ensures the same sequence is produced every time the notebook is run.

---
## Task 3 — Create a Random NumPy Array

We generate a NumPy array called `arr` containing **6 random integers** between 0 (inclusive) and 5 (exclusive), using seed 42 for reproducibility. The array is explicitly stored as `int32` so that the required dtype conversion can be demonstrated in a later task.

```python
# Set the seed so random values are reproducible
set_seed(42)

# 6 random integers in [0, 5)
arr = np.random.randint(0, 5, size=6).astype(np.int32)

print("arr:  ", arr)
print("dtype:", arr.dtype)
print("shape:", arr.shape)
```

---
## Task 4 — Convert `arr` into Tensor `x`

We use `torch.from_numpy()` to create a PyTorch tensor from the NumPy array. This method can share underlying memory with the source array, a concept discussed in the original lesson.

```python
x = torch.from_numpy(arr)

print("x:    ", x)
print("type: ", type(x))
print("dtype:", x.dtype)
```

The tensor `x` inherits the `int32` data type from the NumPy array.

---
## Task 5 — Change the Data Type of `x`

The task requires converting `x` from `int32` to `int64`. This is done using `.type()`, which returns a new tensor with the specified dtype.

```python
print("Before:", x.dtype)

x = x.type(torch.int64)

print("After: ", x.dtype)
```

---
## Task 6 — Reshape `x` into a 3 × 2 Tensor

The tensor `x` contains 6 elements, and $3 \times 2 = 6$, so reshaping into a 3×2 tensor is valid. Reshaping changes the arrangement of elements without changing the total count.

```python
x = x.reshape(3, 2)

print(x)
print("Shape:", x.shape)
```

---
## Task 7 — Return the Right-Hand Column of `x`

Since `x` now has shape (3, 2), the right-hand column is column index 1. Using `x[:, 1:]` selects all rows and columns from index 1 onward, preserving the 2-D shape as a column tensor.

```python
right_column = x[:, 1:]

print("Right-hand column:")
print(right_column)
print("Shape:", right_column.shape)
```

The `1:` slice keeps the result as a 2-D tensor with shape (3, 1), rather than collapsing it to a 1-D tensor of shape (3,) as `x[:, 1]` would.

---
## Task 8 — Square the Values of `x`

The instruction says "without changing `x`," so we use a non-in-place operation. The `**` operator squares each element independently (element-wise), which is different from matrix multiplication.

```python
x_squared = x ** 2

print("Original x:")
print(x)

print("\nSquared values:")
print(x_squared)
```

Confirmed: `x` remains unchanged after the squaring operation.

---
## Task 9 — Create Tensor `y` for Matrix Multiplication

Tensor `y` must satisfy four constraints:

1. Contains the **same number of elements** as `x` (which is 6).
2. Created using **PyTorch** (not NumPy).
3. Contains random integers from **0 inclusive to 5 exclusive**.
4. Its shape must permit **matrix multiplication** with `x`.

### Shape Reasoning

For matrix multiplication $A_{m \times n} \cdot B_{n \times p}$, the inner dimensions must match.

Since $x$ has shape $(3 \times 2)$, the first dimension of $y$ must be **2** (to match the 2 columns of $x$).

Since $y$ must also contain exactly 6 elements:

$$2 \times 3 = 6$$

Therefore $y$ must have shape $(2 \times 3)$, and the result will be:

$$(3 \times 2) \cdot (2 \times 3) = (3 \times 3)$$

```python
# Reuse the seed function for consistency
set_seed(42)

# y must be 2x3 so its rows match x's 2 columns
y = torch.randint(0, 5, (2, 3))

print("y:")
print(y)
print("Shape:             ", y.shape)
print("Number of elements:", y.numel())
```

---
## Task 10 — Find the Matrix Product of `x` and `y`

Before multiplying, we confirm the shapes are compatible.

```python
print("x shape:", x.shape)
print("y shape:", y.shape)
```

$(3 \times 2) \cdot (2 \times 3)$ is valid because the inner dimensions (2) match. The resulting matrix will have shape $(3 \times 3)$.

```python
matrix_product = torch.mm(x, y)

print("Matrix product (x @ y):")
print(matrix_product)
print("Result shape:", matrix_product.shape)
```

We can verify the same result using the `@` operator:

```python
# Equivalent using the @ operator
print(x @ y)
```

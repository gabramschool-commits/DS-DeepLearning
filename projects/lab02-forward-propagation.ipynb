# Laboratory Task 2 — Forward Propagation in a Neural Network

**Understanding Deep Learning**


This notebook performs a **single forward pass** through a small feedforward neural network using the inputs, weights, biases, and ReLU activation function specified by the laboratory instructions.

We will calculate:
- Hidden-layer weighted sums ($Z_1$, $Z_2$)
- Hidden-layer activations ($H_1$, $H_2$)
- Output-layer weighted sum ($Z_3$)
- Predicted output ($\hat{y}$)
- Prediction error ($\delta = \hat{y} - y$)

No training or weight updates are performed here — the goal is to demonstrate the mathematics of one feedforward pass.

## Given Values

### Input

$$x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}$$

### Target

$$y = \begin{bmatrix} 1 \end{bmatrix}$$

### Activation Function — ReLU

$$f(Z) = \max(0, Z)$$

### Hidden Unit Weights

$$w_{11} = 0.2 \qquad w_{12} = -0.3$$

$$w_{13} = 0.4 \qquad w_{14} = 0.1$$

$$w_{15} = -0.5 \qquad w_{16} = 0.2$$

### Output Unit Weights

$$w_{21} = -0.3 \qquad w_{22} = -0.2$$

### Biases

$$\theta_1 = -0.4 \qquad \theta_2 = 0.2 \qquad \theta_3 = 0.1$$

## Network Structure

The network has the following architecture:

```
3 Input Neurons  (x1, x2, x3)
       ↓
2 Hidden Neurons (H1, H2)
       ↓
1 Output Neuron  (ŷ)
```

The connections follow the weight mapping from the original lesson:

- **Hidden Neuron 1** receives inputs through weights $w_{11}$, $w_{13}$, $w_{15}$ and bias $\theta_1$.
- **Hidden Neuron 2** receives inputs through weights $w_{12}$, $w_{14}$, $w_{16}$ and bias $\theta_2$.
- **Output Neuron** receives hidden activations through weights $w_{21}$, $w_{22}$ and bias $\theta_3$.

## Step 1 — Hidden Neuron 1

The weighted sum for hidden neuron 1 is:

$$Z_1 = (w_{11} \cdot x_1) + (w_{13} \cdot x_2) + (w_{15} \cdot x_3) + \theta_1$$

Substituting the given values:

$$Z_1 = (0.2 \times 1) + (0.4 \times 0) + (-0.5 \times 1) + (-0.4)$$

$$Z_1 = 0.2 + 0.0 + (-0.5) + (-0.4)$$

$$Z_1 = 0.2 - 0.5 - 0.4$$

$$\boxed{Z_1 = -0.7}$$

Now apply the ReLU activation:

$$H_1 = ReLU(Z_1) = \max(0,\; -0.7)$$

$$\boxed{H_1 = 0.0}$$

Since $Z_1$ is negative, applying ReLU sets the activation of hidden neuron 1 to zero. This neuron is **inactive** for this input.

## Step 2 — Hidden Neuron 2

The weighted sum for hidden neuron 2 is:

$$Z_2 = (w_{12} \cdot x_1) + (w_{14} \cdot x_2) + (w_{16} \cdot x_3) + \theta_2$$

Substituting the given values:

$$Z_2 = (-0.3 \times 1) + (0.1 \times 0) + (0.2 \times 1) + 0.2$$

$$Z_2 = -0.3 + 0.0 + 0.2 + 0.2$$

$$\boxed{Z_2 = 0.1}$$

Now apply the ReLU activation:

$$H_2 = ReLU(Z_2) = \max(0,\; 0.1)$$

$$\boxed{H_2 = 0.1}$$

Since $Z_2$ is positive, hidden neuron 2 remains **active** with an activation value of 0.1.

## Step 3 — Output Neuron

The weighted sum for the output neuron uses the hidden-layer activations:

$$Z_3 = (w_{21} \cdot H_1) + (w_{22} \cdot H_2) + \theta_3$$

Substituting:

$$Z_3 = (-0.3 \times 0.0) + (-0.2 \times 0.1) + 0.1$$

$$Z_3 = 0.0 + (-0.02) + 0.1$$

$$\boxed{Z_3 = 0.08}$$

Now apply ReLU to produce the final prediction:

$$\hat{y} = ReLU(Z_3) = \max(0,\; 0.08)$$

$$\boxed{\hat{y} = 0.08}$$

## Step 4 — Calculate the Error

The output error is the difference between the predicted output and the actual target, as described in the original lesson:

$$\delta = \hat{y} - y$$

$$\delta = 0.08 - 1$$

$$\boxed{\delta = -0.92}$$

The negative sign indicates that the network **under-predicted** the target — the actual value is higher than what the network produced.

For reference, the squared error can also be computed as an additional loss measure:

$$(\hat{y} - y)^2 = (-0.92)^2 = 0.8464$$

This squared error is commonly used as a loss function in neural network training, where larger errors are penalized more heavily.

## Python Verification

The following NumPy code reproduces the manual calculations above to confirm correctness.

````python
import numpy as np

# --- Given Values ---
x = np.array([1.0, 0.0, 1.0])
y = np.array([1.0])

# ReLU activation function
def relu(z):
    return np.maximum(0, z)

# Hidden unit weights
w11, w12 = 0.2, -0.3
w13, w14 = 0.4,  0.1
w15, w16 = -0.5, 0.2

# Output unit weights
w21 = -0.3
w22 = -0.2

# Biases
theta1 = -0.4
theta2 =  0.2
theta3 =  0.1

# --- Hidden Layer ---
Z1 = (w11 * x[0]) + (w13 * x[1]) + (w15 * x[2]) + theta1
H1 = relu(Z1)

Z2 = (w12 * x[0]) + (w14 * x[1]) + (w16 * x[2]) + theta2
H2 = relu(Z2)

# --- Output Layer ---
Z3 = (w21 * H1) + (w22 * H2) + theta3
y_hat = relu(Z3)

# --- Error ---
error = y_hat - y[0]
squared_error = error ** 2

# --- Print Results ---
print(f"Z1            = {Z1}")
print(f"H1 = ReLU(Z1) = {H1}")
print(f"Z2            = {Z2}")
print(f"H2 = ReLU(Z2) = {H2}")
print(f"Z3            = {Z3}")
print(f"y_hat         = {y_hat}")
print(f"Error         = {error}")
print(f"Squared Error = {squared_error}")
````

## Final Result Table

| Quantity | Result |
| :--- | ---: |
| $Z_1$ | $-0.7$ |
| $H_1 = ReLU(Z_1)$ | $0.0$ |
| $Z_2$ | $0.1$ |
| $H_2 = ReLU(Z_2)$ | $0.1$ |
| $Z_3$ | $0.08$ |
| $\hat{y} = ReLU(Z_3)$ | $0.08$ |
| Error $(\hat{y} - y)$ | $-0.92$ |
| Squared Error $(\hat{y} - y)^2$ | $0.8464$ |

## Interpretation

After applying ReLU, only hidden neuron 2 remained active ($H_2 = 0.1$), while hidden neuron 1 was deactivated ($H_1 = 0.0$) because its weighted sum was negative ($Z_1 = -0.7$). This means the output neuron received a meaningful signal from only one of the two hidden neurons.

The network produced a prediction of $\hat{y} = 0.08$, which is far from the target value of $y = 1$. The error $\delta = -0.92$ confirms that the network significantly under-predicted the target. This is expected because the weights and biases were not trained — they were simply given initial values. To reduce this error, the network would need to adjust its weights and biases using a learning algorithm such as **backpropagation**, which is the focus of Laboratory Task 3.
